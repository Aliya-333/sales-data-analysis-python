## Missing Value Analysis

The dataset contains missing values in several columns. Instead of
immediately removing these rows, the missing values were investigated
because the sales register contains different types of records.

Missing values may represent the structure of the original sales report
rather than incorrect or incomplete data.

```python
print("Total Rows:", len(df))
```

```python
products = df[
    df["Quantity"].notna() &
    df["Rate"].notna() &
    df["Value_Without_Tax"].notna()
].copy()
```

```python
print("Original Rows:", len(df))
print("Product Rows:", len(products))
```

```python
products.head(10)
```

```python
products.isnull().sum()
```

```python
products = products[
    ["Item", "Quantity", "Rate", "Value_Without_Tax"]
].copy()

products.head()
```

```python
products["Item"] = products["Item"].str.strip()
```

```python
products.info()
```

## Data Cleaning

The original dataset contains 2,149 rows. Several columns contain
missing values because the sales register includes different types
of records.

For product-level sales analysis, records containing valid Quantity,
Rate, and Value Without Tax values were selected.

This created a separate product-level dataset that can be used for
sales and quantity analysis without removing useful records from the
original dataset.

```python
print("Duplicate rows in original data:", df.duplicated().sum())
print("Duplicate rows in product data:", products.duplicated().sum())
```

```python
products.describe()
```

```python
print("Total Quantity Sold:", products["Quantity"].sum())
print("Total Sales Value:", products["Value_Without_Tax"].sum())
print("Average Rate:", products["Rate"].mean())
print("Highest Rate:", products["Rate"].max())
print("Lowest Rate:", products["Rate"].min())
```

```python
unique_products = products["Item"].nunique()

print("Number of Unique Products:", unique_products)
```

```python
print(products["Item"].unique())
```

```python
top_sales = (
    products.groupby("Item")["Value_Without_Tax"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_sales)
```

```python
top_quantity = (
    products.groupby("Item")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_quantity)
```

```python
product_summary = products.groupby("Item").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Value_Without_Tax", "sum"),
    Average_Rate=("Rate", "mean")
)

product_summary = product_summary.sort_values(
    by="Total_Sales",
    ascending=False
)

product_summary.head(10)
```

```python
# Find highest sales-value product
highest_sales_product = product_summary.iloc[0]

print("Highest Sales Product:")
print(highest_sales_product)
print("Product:", product_summary.index[0])
```

```python
# Find highest quantity-selling product
quantity_summary = products.groupby("Item")["Quantity"].sum()

top_quantity_product = quantity_summary.idxmax()
top_quantity_value = quantity_summary.max()

print("Top Selling Product:", top_quantity_product)
print("Quantity Sold:", top_quantity_value)
```

```python
highest_transaction = products.loc[
    products["Value_Without_Tax"].idxmax()
]

print(highest_transaction)
```

```python
invoices = df[
    df["Date"].notna() &
    df["Gross_Total"].notna()
].copy()

print("Invoice Records:", len(invoices))
```

```python
invoices.head()
```

```python
gross_sales = invoices["Gross_Total"].sum()

print("Total Gross Sales:", gross_sales)
```

```python
total_cgst = invoices["CGST"].sum()
total_sgst = invoices["SGST"].sum()

print("Total CGST:", total_cgst)
print("Total SGST:", total_sgst)
print("Total Tax:", total_cgst + total_sgst)
```

```python
print(invoices["Payment_Terms"].value_counts())
```

```python
invoices["Payment_Terms"] = (
    invoices["Payment_Terms"]
    .astype("string")
    .str.strip()
    .str.upper()
)
```

```python
print(invoices["Payment_Terms"].value_counts())
```

```python
# Analyze sales by date

# First make sure Date is a proper date:

invoices["Date"] = pd.to_datetime(
    invoices["Date"],
    errors="coerce"
)
```

```python
invoices[["Date", "Gross_Total"]].head()
```

```python
daily_sales = (
    invoices.groupby("Date")["Gross_Total"]
    .sum()
    .sort_index()
)

print(daily_sales.head(10))
```

```python
# Create a histogram showing the distribution of product rates:
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file
df = pd.read_csv("cleaned_sales_data.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Check column names
print(df.columns.tolist())

# Create histogram
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Rate",
    bins=30,
    kde=True
)

plt.title("Distribution of Product Rates")
plt.xlabel("Rate")
plt.ylabel("Frequency")

plt.show()
```

```python
# Pie Chart + Scatter Plot

# For a pie chart, use the top products by quantity:

top_quantity = (
    df.groupby("Item")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

plt.figure(figsize=(8, 8))

plt.pie(
    top_quantity.values,
    labels=top_quantity.index,
    autopct="%1.1f%%"
)
plt.title("Top 5 Products by Quantity Sold")
plt.show()
```

```python
# scatter plot:

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Value_Without_Tax"
)

plt.title("Quantity vs Sales Value")
plt.xlabel("Quantity")
plt.ylabel("Sales Value")
plt.show()
```

```python
# Daily sales chart
daily_sales.plot(
    kind="line",
    figsize=(12, 5)
)

plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Gross Sales")
plt.tight_layout()
plt.show()
```
